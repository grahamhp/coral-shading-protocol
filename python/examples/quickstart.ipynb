# Coral Shading Protocol v5: quick start

The same calculation as the browser app, in Python. Light data come from PVGIS (downloaded and cached) or the bundled Sipaway example.

```python
from coralshading import design

r = design.design(10.4675, 123.4506, tz=8, start=(2025, 4, 28), end=(2025, 8, 8),
                  target="reef_star", depth=10, height=0.75, kd=0.10, w0=0.7,
                  scenario="both", sync=True, cloth_rating=80, coral="typical",
                  light="example", spacing=0.1)   # light="tmy" downloads a PVGIS typical year
print(r.summary())
```

## Daily light across the coral area

```python
import matplotlib.pyplot as plt
fig, axes = plt.subplots(1, len(r.scenarios), figsize=(10, 4))
for ax, s in zip(axes, r.scenarios):
    sc = ax.scatter(s.X, s.Y, c=s.period.dli().mean(axis=0), s=40, marker='s', cmap='viridis')
    ax.set_title(f"{s.kind}: {s.panel.lx:.2f} x {s.panel.ly:.2f} m"); ax.set_aspect('equal')
    fig.colorbar(sc, ax=ax, label='daily light (mol m$^{-2}$ d$^{-1}$)')
fig.tight_layout()
```
